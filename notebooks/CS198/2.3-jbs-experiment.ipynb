```python
# -------------------------------------
# Library configuration
# -------------------------------------
# Standard libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
# -------------------------------------
# Settings
# -------------------------------------
random_state = 42
```

## Data preprocessing

```python
# Loading dataset
df = pd.read_csv('data/data.csv')
```

```python
# Displaying dataset
df
```

```python
# Sorting dataset based on SAMPLEID
df_sorted = df.sort_values(by=['SAMPLEID'])
```

```python
# Displaying sorted dataset
df_sorted
```

```python
# Dropping empty rows from sorted dataset
df_cleaned = df_sorted.dropna()
```

```python
# Displaying cleaned dataset
df_cleaned
```

```python
# Gathering column names
column_names = df_cleaned.columns.tolist()
```

```python
# Displaying column names
column_names
```

```python
# Setting targets
df_targets = df_cleaned[['6_K6_total',
 '6_PHQ9_total',
 '6_GAD7_total',
 '6_SSS8_total',
 '6_PTGI-X(Q7_22_27)',
 '6_SHS_total',
 '6_UCLA_total',
 '6_LSNS6_total',
 '6_AUDIT']]

num_targets = len(df_targets.columns.tolist())
```

```python
# Setting features
df_features = df_cleaned.drop([
 'SAMPLEID',
 'ANSWERDATE',
 '6_K6_total',
 '6_PHQ9_total',
 '6_GAD7_total',
 '6_SSS8_total',
 '6_PTGI-X(Q7_22_27)',
 '6_SHS_total',
 '6_UCLA_total',
 '6_LSNS6_total',
 '6_AUDIT'], axis=1)

num_features = len(df_features.columns.tolist())
```

```python
df_features
```

```python
# Feature scaling
# Used as reference: https://analyticsindiamag.com/ai-mysteries/a-beginners-guide-to-scikit-learns-mlpclassifier/
#for col in df_features.columns.tolist():
#    df_features[col] = LabelEncoder().fit_transform(df_features[col])

for col in df_targets.columns.tolist():
    df_targets[col] = LabelEncoder().fit_transform(df_targets[col])
    
print(df_targets)
```

```python
# Multilabel Binarizer
from sklearn.preprocessing import LabelBinarizer

target_names = df_targets.columns.tolist()
binary_targets = {}

for target in target_names:
 lb = LabelBinarizer()
 binary_targets[target] = lb.fit_transform(df_targets[target])
 print(f"{target}")
 print(binary_targets[target])
```

```python
# Make target arrays based on the following ranges
# K6 - 0 to 24
# PHQ-9 - 0 to 27
# GAD-7 - O to 21
# SSS-8 - 0 to 32
# LSNS - 0 to 60
# UCLA - 0 to 80
# SHS - 1 to 7

binary_targets_df = pd.concat(
    [pd.DataFrame(binary_targets[target], columns=[f"{target}_{i}" for i in range(binary_targets[target].shape[1])]) 
     for target in binary_targets],
    axis=1
)

print(binary_targets_df)
```

# Building the model

We pivot to using scikit-learn because of the finicky nature of PyTorch

https://scikit-learn.org/dev/modules/generated/sklearn.neural_network.MLPClassifier.html

Since we see that the number of targets is greater than one and the cardinality is greater than two, we then know that we have a multiclass-multioutput problem

Reference: https://scikit-learn.org/1.5/modules/multiclass.html

```python
# -------------------------------------
# Library configuration
# -------------------------------------
# Standard libraries
from sklearn.neural_network import MLPClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
```

```python
X = df_features
y = binary_targets_df

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=random_state)
```

```python
clf = MLPClassifier(max_iter=5000, random_state=random_state).fit(X_train, y_train)
```

```python
clf.predict_proba(X_test)
```

```python
clf.predict(X_test)
```

```python
clf.score(X_test, y_test)
```
