```python
import pandas as pd
import datetime as dt
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

phase1 = pd.read_csv('2-interim/phase1.csv')
phase2 = pd.read_csv('2-interim/phase2.csv')
phase3 = pd.read_csv('2-interim/phase3.csv')
phase4 = pd.read_csv('2-interim/phase4.csv')
```

```python
# https://saturncloud.io/blog/how-to-select-date-range-from-pandas-dataframe/
date_ranges = [
    ('2020-05-11', '2020-05-12'),  # Phase 1, n = 6723
    ('2021-06-14', '2021-06-20'),  # Phase 2, n = 4592
    ('2022-05-13', '2022-05-30'),  # Phase 3, n = 3892
    ('2024-05-10', '2024-05-17')  # Phase 4, n = 2659
]
```

```python
prefectures = [
    13,  # Tokyo 
    27,  # Osaka
    28,  # Hyogo
    40  # Fukuoka
]
```

```python
expected_shapes = [
    6723,
    4592,
    3892,
    2659
]
```

```python
# Prefecture columns and answer columns for each phase
prefecture_cols = ['1_PREFECTURE', '3_PREFECTURE', '5_PREFECTURE', '6_PREFECTURE']
answer_cols = ['1_ANSWERDATE', '3_ANSWERDATE', 'ANSWERDATE', '6_ANSWERDATE']
```

```python
def verify_phase_data(phase, phase_num):
    # Convert to datetime format
    phase[answer_cols[phase_num]] = pd.to_datetime(phase[answer_cols[phase_num]]).dt.date
    phase[answer_cols[phase_num]] = pd.to_datetime(phase[answer_cols[phase_num]])
    
    start_date = date_ranges[phase_num][0]
    end_date = date_ranges[phase_num][1]

    # Check if dataframe is in the expected date range
    all_in_range = phase[answer_cols[phase_num]].between(start_date, end_date).all()
    
    # Check if dataframe has the expected number of entries
    phase_shape = phase.index.size == expected_shapes[phase_num]
    
    print(f"Phase {phase_num + 1}:\n"
          f"All dates in range ({date_ranges[phase_num]})? -> {all_in_range}\n"
          f"Expected shape ({expected_shapes[phase_num]})? -> {phase_shape}")
```

```python
phase1_filtered = phase1[phase1['1_PREFECTURE'].isin(prefectures)]
phase1_filtered = phase1_filtered.dropna(subset=['1_ANSWERDATE'])
verify_phase_data(phase1_filtered, 0)
phase1_filtered.columns = phase1_filtered.columns.str.replace(r'^\d+_', '', regex=True).str.lower()
phase1_filtered
```

```python
# Export to CSV
phase1_filtered.to_csv("3-processed/phase1.csv", index=False)
```

```python
phase2_filtered = phase2[phase2['3_PREFECTURE'].isin(prefectures)]
phase2_filtered = phase2[phase2['SAMPLEID'].isin(phase1_filtered['sampleid'])]
phase2_filtered = phase2_filtered.dropna(subset=['3_ANSWERDATE'])
verify_phase_data(phase2_filtered, 1)
phase2_filtered.columns = phase2_filtered.columns.str.replace(r'^\d+_', '', regex=True).str.lower()
phase2_filtered
```

```python
# Export to CSV
phase2_filtered.to_csv("3-processed/phase2.csv", index=False)
```

```python
phase3_filtered = phase3[phase3['5_PREFECTURE'].isin(prefectures)]
phase3_filtered = phase3[phase3['SAMPLEID'].isin(phase2_filtered['sampleid'])]
phase3_filtered = phase3_filtered.dropna(subset=['ANSWERDATE'])
verify_phase_data(phase3_filtered, 2)
phase3_filtered.columns = phase3_filtered.columns.str.replace(r'^\d+_', '', regex=True).str.lower()
phase3_filtered
```

```python
# Export to CSV
phase3_filtered.to_csv("3-processed/phase3.csv", index=False)
```

```python
phase4_filtered = phase4[phase4['6_PREFECTURE'].isin(prefectures)]
phase4_filtered = phase4[phase4['SAMPLEID'].isin(phase3_filtered['sampleid'])]
phase4_filtered = phase4.dropna(subset=['6_ANSWERDATE'])
verify_phase_data(phase4_filtered, 3)
phase4_filtered.columns = phase4_filtered.columns.str.replace(r'^\d+_', '', regex=True).str.lower()
phase4_filtered
```
