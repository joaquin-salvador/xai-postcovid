# Classification

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import KBinsDiscretizer
import matplotlib.pyplot as plt
import numpy as np

# Ensure y_train and y_test are numpy arrays
y_train_array = y_train.to_numpy() if isinstance(y_train, pd.Series) else y_train
y_test_array = y_test.to_numpy() if isinstance(y_test, pd.Series) else y_test

# 1. Bin the target into 4 balanced classes using quantiles
kbins = KBinsDiscretizer(n_bins=2, encode='ordinal', strategy='quantile')

y_train_binned = kbins.fit_transform(y_train_array.reshape(-1, 1)).astype(int).ravel()
y_test_binned = kbins.transform(y_test_array.reshape(-1, 1)).astype(int).ravel()
```

```python
# 2. Train classifier
clf = RandomForestClassifier(random_state=random_state)
clf.fit(X_train_scaled, y_train_binned)
```

```python
# 3. Evaluation function for classification
def evaluate_classifier(model, X_test, y_true):
    y_pred = model.predict(X_test)
    
    # Confusion matrix
    cm = confusion_matrix(y_true, y_pred)
    print("Confusion Matrix:\n", cm)
    
    # Classification report
    print("\nClassification Report:\n", classification_report(y_true, y_pred))
    
    # Scatter plot of predictions
    plt.figure(figsize=(6, 4))
    plt.scatter(y_true, y_pred, alpha=0.4, c=y_pred, cmap='viridis')
    plt.xlabel("Actual Class")
    plt.ylabel("Predicted Class")
    plt.title("Predicted vs Actual Classes")
    plt.colorbar(label="Predicted Class")
    plt.show()
```

```python
# 4. Evaluate
evaluate_classifier(clf, X_test_scaled, y_test_binned)
```

```python
# 5. Plot class distribution
def plot_class_distribution(y, title):
    plt.hist(y, bins=np.arange(5) - 0.5, rwidth=0.8, align='mid')
    plt.xticks([0, 1, 2, 3])
    plt.xlabel("Class")
    plt.ylabel("Frequency")
    plt.title(title)
    plt.grid(axis='y')
    plt.show()

plot_class_distribution(y_train_binned, "Training Class Distribution")
plot_class_distribution(y_test_binned, "Test Class Distribution")
```

```python
baseline_model = RandomForestRegressor(random_state=random_state, n_estimators=1000)
baseline_model.fit(X_train_scaled, y_train_scaled.ravel())
```

```python
def evaluate(model, X_test_scaled, y_test):
    y_pred_scaled = model.predict(X_test_scaled)
    
    y_pred = scaler_y.inverse_transform(y_pred_scaled.reshape(-1, 1))
    
    plt.scatter(y_test, y_pred, alpha=0.5)
    plt.xlabel("Actual Values")
    plt.ylabel("Predicted Values")
    plt.title("Predicted vs Actual Values")
    plt.show()
    
    # Metrics
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    print(f"{model.__class__.__name__}:")
    print(f"  MAE: {mae:.4f}")
    print(f"  MSE: {mse:.4f}")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  R²: {r2:.4f}")
    print("-" * 30)
    
    return r2
```

```python
evaluate(baseline_model, X_test_scaled, y_test)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import BatchNormalization
```

```python
keras_model = Sequential([
    Dense(128, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    BatchNormalization(),
    Dropout(0.2),
    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1)
])
```

```python
from tensorflow.keras.optimizers.schedules import ExponentialDecay

lr_schedule = ExponentialDecay(
    initial_learning_rate=0.001,
    decay_steps=100,
    decay_rate=0.9)

optimizer = Adam(learning_rate=lr_schedule)
keras_model.compile(optimizer=optimizer, loss='mse', metrics=['mae'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

checkpoint = ModelCheckpoint("best_model.keras", monitor="val_loss", save_best_only=True)

history = keras_model.fit(
    X_train_scaled, y_train_scaled,
    validation_split=0.1,
    epochs=1000,
    batch_size=32,
    callbacks=[early_stop, checkpoint],
    verbose=1
)
```

```python
def evaluate_keras(model, X_test_scaled, y_test):
    y_pred_scaled = model.predict(X_test_scaled)
    y_pred = scaler_y.inverse_transform(y_pred_scaled)

    plt.scatter(y_test, y_pred, alpha=0.5)
    plt.xlabel("Actual Values")
    plt.ylabel("Predicted Values")
    plt.title("Predicted vs Actual Values")
    plt.show()

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    print(f"TensorFlow Model:")
    print(f"  MAE: {mae:.4f}")
    print(f"  MSE: {mse:.4f}")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  R²: {r2:.4f}")
    print("-" * 30)

    return r2
```

```python
evaluate_keras(keras_model, X_test_scaled, y_test)
```

# With classification loss

```python
import numpy as np

n_classes = 4

# Combine train and test for consistent binning
all_targets = np.concatenate([y_train_scaled, y_test_scaled])

# Define bin edges based on percentiles or linspace
bin_edges = np.linspace(all_targets.min(), all_targets.max(), n_classes + 1)

# Digitize the targets into class labels
y_train_class = np.digitize(y_train_scaled, bins=bin_edges[1:-1])  # exclude outer edges
y_test_class = np.digitize(y_test_scaled, bins=bin_edges[1:-1])
```

```python
from tensorflow.keras.utils import to_categorical

y_train_class = to_categorical(y_train_class, num_classes=n_classes)
y_test_class = to_categorical(y_test_class, num_classes=n_classes)
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization

# Input layer
input_layer = Input(shape=(X_train_scaled.shape[1],))

# Shared layers
x = Dense(128, activation='relu')(input_layer)
x = BatchNormalization()(x)
x = Dropout(0.2)(x)
x = Dense(64, activation='relu')(x)
x = BatchNormalization()(x)
x = Dropout(0.2)(x)
x = Dense(32, activation='relu')(x)

# Regression output
regression_output = Dense(1, name="regression")(x)

# Classification output (e.g., into 10 bins/classes)
num_classes = 4
classification_output = Dense(num_classes, activation="softmax", name="classification")(x)

# Model with 2 outputs
model = Model(inputs=input_layer, outputs=[regression_output, classification_output])
```

```python
import numpy as np

# Define class bins for classification loss
bins = np.linspace(y_train_scaled.min(), y_train_scaled.max(), num_classes + 1)
y_train_class = np.digitize(y_train_scaled, bins) - 1
y_train_class = np.clip(y_train_class, 0, num_classes - 1)
```

```python
# Create a new optimizer instance
from tensorflow.keras.optimizers.schedules import ExponentialDecay
from tensorflow.keras.optimizers import Adam

lr_schedule = ExponentialDecay(
    initial_learning_rate=0.001,
    decay_steps=100,
    decay_rate=0.9
)
new_optimizer = Adam(learning_rate=lr_schedule)

# Recompile the model with the new optimizer
model.compile(
    optimizer=new_optimizer,
    loss={
        "regression": "mse",
        "classification": "sparse_categorical_crossentropy"
    },
    loss_weights={
        "regression": 1.0,
        "classification": 1.0  # tune this λ
    },
    metrics={"regression": "mae", "classification": "accuracy"}
)
```

```python
#import tensorflow as tf
#tf.config.run_functions_eagerly(True)

history = model.fit(
    X_train_scaled,
    {"regression": y_train_scaled, "classification": y_train_class},
    validation_split=0.1,
    epochs=1000,
    batch_size=32,
    callbacks=[early_stop, checkpoint],
    verbose=1
)
```

# Model Tuning

```python
import pprint
import joblib as job
import time

from sklearn.model_selection import GridSearchCV

pp = pprint.PrettyPrinter(indent=1)
```

```python
# Define hyperparameter space
n_estimators = [100, 250, 500, 1000]                       # Range of trees
max_features = [1, 2, 3, 4]                                # Feature selection
max_depth = [int(x) for x in np.linspace(10, 100, num=10)]
max_depth.append(None) # Depth
min_samples_split = [2, 5, 10, 15]                     # Control overfitting
min_samples_leaf = [1, 2, 4, 6]                       # Prevent overfitting
bootstrap = [True, False]                              # Sampling strategy
cv = 10                                         # Cross-validation folds

# Create the random grid
random_grid = {'n_estimators': n_estimators,
               'max_features': max_features,
               'max_depth': max_depth,
               'min_samples_split': min_samples_split,
               'min_samples_leaf': min_samples_leaf,
               'bootstrap': bootstrap}

print('Hyperparameter search space')
pp.pprint(random_grid)
```

```python
# Set model to RandomForestRegressor
model = RandomForestRegressor(random_state=random_state)
```

```python
start_time = time.time()
grid_tuner = GridSearchCV(estimator=model, param_grid=random_grid, cv=cv, scoring='r2', 
                     verbose=6, return_train_score=True, n_jobs=10)
print('\nCross-validation via', grid_tuner)

# Fit the random search model
print("Fitting the Grid Search model...")
grid_tuner.fit(X_train_scaled, y_train_scaled.ravel())
print("Grid Search CV completed.")

end_time = time.time()
running_time = end_time - start_time
print('\nRandom Forest training time:', running_time, 's')
```

```python
## Plot CV results
def plot_cv(tuner):
    results = tuner.cv_results_
    cv_params = tuner.param_grid
    means_test = results['mean_test_score']
    stds_test = results['std_test_score']
    means_train = results['mean_train_score']
    stds_train = results['std_train_score']

    print(cv_params)
    fig, ax = plt.subplots(1, len(cv_params), sharex='none', sharey='all', figsize=(20,8), dpi=300)
    fig.suptitle('Hyperparameter Tuning')
    fig.text(0.04, 0.5, 'MEAN SCORE', va='center', rotation='vertical')

    scores = {}
    for i, (param, param_val) in enumerate(cv_params.items()):
        x, y1, e1, y2, e2 = [], [], [], [], []

        for v in param_val:
            mask_idx = results['param_'+param]==v
            v = str(v)

            m_train = np.mean(means_train[mask_idx])
            s_train = np.sqrt(np.mean(stds_train[mask_idx]**2))
            m_test = np.mean(means_test[mask_idx])    
            s_test = np.sqrt(np.mean(stds_test[mask_idx]**2))

            scores[param+'='+v+'_means_train'] = m_train
            scores[param+'='+v+'_stds_train'] = s_train    
            scores[param+'='+v+'_means_test'] = m_test
            scores[param+'='+v+'_stds_test'] = s_test
            print(param, "=", v, ":", m_test)

            x.append(v)
            y1.append(m_train)
            e1.append(s_train) 
            y2.append(m_test) 
            e2.append(s_test)

        ax[i].errorbar(x, y1, e1, linestyle='-', marker='^', label='Train')
        ax[i].errorbar(x, y2, e2, linestyle='--', marker='o', color='g', label='CV')
        ax[i].set_xlabel(param.upper())

    pp.pprint(scores)

    plt.legend()
    plt.show()

sns.set_style('dark')
plot_cv(saved_model)
print(tuner.cv_results_)
```
